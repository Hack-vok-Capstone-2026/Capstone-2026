```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\Harshan A\Downloads\Capstone-2026\data\raw\processed\Data_cleaned.csv")

print("Shape:", df.shape)
```

```python
member3_columns = [
    "koi_prad",
    "koi_period",
    "koi_depth",
    "koi_teq",
    "koi_insol",
    "koi_steff",
    "koi_slogg",
    "koi_srad",
    "koi_duration",
    "koi_impact",
    "koi_model_snr",
    "koi_kepmag"
]
print(df["koi_prad"].describe())

plt.figure(figsize=(8, 5))
plt.hist(df["koi_prad"].dropna(), bins=50)
plt.xlabel("Planet Radius (Earth Radii)")
plt.ylabel("Number of Candidates")
plt.title("Planet Radius Distribution")
plt.show()
```

```python
print(df["koi_period"].describe())

plt.figure(figsize=(8, 5))
plt.hist(df["koi_period"].dropna(), bins=50)
plt.xlabel("Orbital Period (Days)")
plt.ylabel("Number of Candidates")
plt.title("Orbital Period Distribution")
plt.show()
```

```python
print(df["koi_depth"].describe())

plt.figure(figsize=(8, 5))
plt.hist(df["koi_depth"].dropna(), bins=50)
plt.xlabel("Transit Depth (ppm)")
plt.ylabel("Number of Candidates")
plt.title("Transit Depth Distribution")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["koi_prad"],
    df["koi_depth"],
    alpha=0.5
)

plt.xlabel("Planet Radius (Earth Radii)")
plt.ylabel("Transit Depth (ppm)")
plt.title("Planet Radius vs Transit Depth")
plt.show()

print(
    "Correlation:",
    df["koi_prad"].corr(df["koi_depth"])
)
```

```python
earth_sized = df[
    (df["koi_prad"] >= 0.8) &
    (df["koi_prad"] <= 1.5)
]

print("Earth-sized candidates:", len(earth_sized))

print(
    earth_sized[
        ["kepoi_name", "koi_prad", "koi_period", "koi_depth", "koi_teq"]
    ].head(20)
)
```

```python
temperate = df[
    (df["koi_teq"] >= 180) &
    (df["koi_teq"] <= 350)
]

print("Potentially temperate candidates:", len(temperate))
```

```python
screened_candidates = df[
    (df["koi_prad"] >= 0.8) &
    (df["koi_prad"] <= 1.5) &
    (df["koi_teq"] >= 180) &
    (df["koi_teq"] <= 350) &
    (df["koi_period"] >= 20)
]

print("Final screened candidates:", len(screened_candidates))

print(
    screened_candidates[
        [
            "kepoi_name",
            "koi_prad",
            "koi_period",
            "koi_depth",
            "koi_teq",
            "koi_insol",
            "koi_steff",
            "koi_srad"
        ]
    ].head(30)
)
```

```python
print("Total candidates:", len(df))
print("Earth-sized candidates:", len(earth_sized))
print("Potentially temperate candidates:", len(temperate))
print("Final screened candidates:", len(screened_candidates))

print("\nMean Planet Radius:", df["koi_prad"].mean())
print("Mean Orbital Period:", df["koi_period"].mean())
print("Mean Transit Depth:", df["koi_depth"].mean())
print("Mean Host-Star Temperature:", df["koi_steff"].mean())
print("Mean Equilibrium Temperature:", df["koi_teq"].mean())

```
